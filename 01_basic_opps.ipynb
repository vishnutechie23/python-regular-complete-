#### OOP's : Object oriented programing 
#### class : blueprint of object (not real entity)
#### object : instance of class (real entity)

```python
class Car:
    #variables
    brand = "Toyota"
    wheels = 4

    #functions
    def accelerate(self):
        print("accelerate")

    def brake(self):
        print("brake")

car1 = Car()  #real car1 object
print(car1.brand)
print(car1.wheels)
car1.accelerate()
car1.brake()

car2 = Car()  #real car1 object
print(car2.brand)
print(car2.wheels)
car2.accelerate()
car2.brake()
```

# Class variables vs instance variables 
### Class variables : 1. shared by all objects 2. Defined outside mehtods
### Instance variables : 1. Unique for each object 2. Deined using "self" 
#### Constructor - Construct Object 
#### A constructor is a special method that is automatically called when an objetc is created 
#### it is used to initialized object data __init__()


```python
class Student:
    def __init__(self, name, rollNo, marks):
        self.name = name  #s1.name = "vishnu"
        self.rollNo = rollNo #s1.rollNo = 1
        self.marks = marks #s1.marks = 100

    def printHello(self):
        print("Hello")

    def printDetails(self):
        print(f"Name : {self.name}")
        print(f"rollNo : {self.rollNo}")
        print(f"Marks : {self.marks}")


s1 = Student("vishnu",1,100)
print(s1.name)
print(s1.rollNo)
print(s1.marks)

s1.printHello()
s1.printDetails()

s2 = Student("aman",2,99)
s2.printDetails()
```

### self : refers to current object and used to access instance variables and methods of the current object
#### How Constructor works interally 
    Object Creation flow :
    1.Memory is allocated for object
    2.__init__() is automatically called
    3.Data is stored inside object

#### Rules
    1.Constructor name is fixed
    2.First parameter is always self
    3.can take multiple parameters

```python
class Employee:
    def __init__(self, name, Id):
        self.name = name
        self.Id = Id
        self.companyName = "TCS"
        self.location = "Mumbai"

e1 = Employee("vishnu dalvi", "12345")
# e1.companyName = "Infosis"
print(e1.name)
print(e1.companyName)
print(e1.location)
```

#### types of constructor 
    1.default constructor
    2.parameterized constructor

```python
class Bankacc:
    def __init__(self, name, id):
        self.name = name
        self.id = id
        self.balance = 0

    def printBalance(self):
        print(self.balance)

acc1 = Bankacc("vishnu", 12345)
acc1.printBalance()
acc1.balance = 10000000 # easily change 
acc1.printBalance()


print(acc1.id)
acc1.id = 54321 # update easily 
print(acc1.id)
```
