```python
# armstrong number : number that is equal to sum of its own digits each equal to a power of number of digits in the numbers 

n = int(input("Enter a number "))

n_str = str(n)
n_digit = len(n_str)

sum_power = 0
temp_n = n 

while temp_n > 0:
    digit = temp_n % 10 
    print("digit:",digit)
    sum_power += digit ** n_digit
    print("sum_power:",sum_power)
    temp_n //= 10 
    print("temp_n:",temp_n)

if sum_power == n:
    print(f"{n} Armstrong number")
else :
    print(f"{n} is not armstrong number")
```

```python
# Binary sorting 
def Binary(nums, target):
    lb = 0
    ub = len(nums) - 1

    while lb <= ub:
        mid = (lb + ub) // 2

        if nums[mid] == target:
            return mid
        elif nums[mid] > target:
            ub = mid - 1
        else:
            lb = mid + 1

    return -1 
```
