1.Introduction to NumPy
1.1 What is NumPy?
-NumPy stands for Numerical Python.

It is a Python library used for:
- Numerical calculations.
- Working with large datasets.
- Mathematical operations.
- Statistical calculations.
- Multidimensional arrays.
- Scientific computing.


```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print(arr)
print(arr + 10)
print(arr * 2)
print("Notice that we performed operations on every element without writing a loop" \
"")
```

```python
# Python list
li = [10, 20, 30, 40, 50]

new_li = [i * 2 for i in li]

print(new_li)
```

```python
# Numpy
arr = np.array([10, 20, 30, 40, 50])

new_arr = arr * 2

print(new_arr)
```

```python
#Python List vs NumPy Array
#Feature	                       Python List             	                    NumPy Array
#Data types	                   Can be mixed	                                Usually homogeneous
#Mathematical operations	       Often require loops	                        Direct element-wise operations
#Memory	                       Generally higher overhead	                Generally more efficient
#Speed	                       Slower for many numerical operations	        Usually faster for numerical workloads
#Dimensions	                   Nested lists	                                Native multidimensional arrays
```

```python
#Execution time comparison
import time
import numpy as np

# Python List
li = list(range(1, 1000001))

start = time.time()

new_li = [i * 2 for i in li]

end = time.time()

print("List Time:", end - start)


# NumPy Array
arr = np.arange(1, 1000001)

start = time.time()

new_arr = arr * 2

end = time.time()

print("NumPy Time:", end - start)
```

```python
# Q6 : Compare execution time between a Python list and NumPy array using 1 million elements.
import time
import numpy as np
# py list
pyli = list(range(1, 1000001))

start = time.time()

new_li = [i * 2 for i in pyli]

end = time.time()

print("List Time:", end - start,"Seconds")

#for np array
arr = np.arange(1, 1000001)

start = time.time()

new_arr = arr * 2

end = time.time()

print("NumPy Time:", end - start,"Seconds")
```

Why is NumPy generally faster?
- Python lists require Python-level iteration for list comprehensions.
- NumPy performs operations using optimized compiled code.
- NumPy supports vectorization, avoiding explicit Python loops.
- One important thing
Don't assume NumPy will always be faster. Performance depends on the operation, data size and implementation.
