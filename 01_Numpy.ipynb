```python
import time # Time Management Library
import sys # Accessing System Resources

import numpy as np # Faster Computation
import random

import warnings
warnings.filterwarnings('ignore')
```

```python
a = 10
b = 20

start = time.time()
ans = a + b
end = time.time()

print(ans)
print(f"Time Taken : {end - start}")
```

```python
# Create a list of 1cr elements and track the time taken to create the list.

start_list = time.time()
li = [item for item in range(1, 20000001)]
end_list = time.time()

start_array = time.time()
arr = np.arange(1, 20000001)
end_array = time.time()

print(f"List\nLength --> {len(li)}\nTime Taken : {end_list - start_list}")
print(f"Array\nLength --> {arr.size}\nTime Taken : {end_array - start_array}")
```

```python
# Memory Taken

print(f"Memory Taken By Python List : {sys.getsizeof(li)}")
print(f"Memory Taken By Numpy Array : {sys.getsizeof(arr)}")
```

```python
# Broadcasting Operations

li = [item for item in range(1, 11, 1)]
print(f"Original List : {li}")

print(f"\nOperation : Add 1 to each element. ")

new_list = [item + 1 for item in li]
print(f"Updated List : {new_list}")
```

```python
li1 = [item for item in range(1, 6, 1)]
li2 = [item for item in range(6, 11, 1)]

concatenated_list = li1 + li2
point_wise_addition_list = [li1[i] + li2[i] for i in range(len(li1))]

print(f"List 1 : {li1}\nList 2 : {li2}\nConcatenated List : {concatenated_list}\nPoint Wise Addition List : {point_wise_addition_list}")
```

```python
arr = np.arange(1, 11, 1)
print(f"Original Array : {arr}")

arr = arr + 1
print(f"Updated Array : {arr}")
```

```python
arr1 = np.arange(1, 6, 1)
arr2 = np.arange(6, 11, 1)

updated_array = arr1 + arr2

print(f"Point Wise Addition Array : {updated_array}")
```

```python
# Data Structures in Numpy - Numpy's Array
# Standard function - np.array()

'''
Ways to Create Array in numpy

1. List to array
2. Tuple to array
3. Dictionary to array
4. Set to array
'''
```

```python
# List to array

li = [int(item) for item in np.random.randint(1, 10, 10)]
print(f"Generated List : {li}")

array = np.array(li)
print(f"Generated Array : {array}")

print(f"Length of List : {len(li)}")
print(f"Length of Array : {array.size}")
```

```python
rows = 3
columns = 5

li = [int(item) for item in np.random.randint(1, 100, rows * columns)]
nested_list = [[li[i * columns + j] for j in range(columns)] for i in range(rows)]

print(f"Original List : {li}")
print(f"Nested List : {nested_list}")
```

```python
# Tuple to array

t = (2345, 234, 34, 234, 2, 7, 5, 7, 5, 3)
print(f"Tuple : {t}\nData Type : {type(t)}")

array = np.array(t)
print(f"\nArray : {array}\nData Type : {type(array)}")

print(f"Length of Tuple : {len(t)}")
print(f'Length of Array : {array.size}')
```

```python
d = {
    "Virat" : 18,
    "ABD" : 17,
    "Chris" : 333
}

print(f"Keys : {d.keys()}")
print(f"Values : {d.values()}")
print(f"Items : {d.items()}")
```

```python
print(f"Dictionary\n{d}\nData Type : {type(d)}\n")

keys_array = np.array(d.keys())
values_array = np.array(d.values())
items_array = np.array(d.items())

print(f"{type(keys_array)}\n{type(values_array)}\n{type(items_array)}")

print(f"Length of Dictionary : {len(d)}")
print(f"Length of Array : {keys_array.size}")
```

```python
s = set((1, 2, 3, 4, 5))
print(f"Set : {s}\nData Type : {type(s)}")

# Set is UnOrdered
# No Repetition Allowed

array = np.array(s)
print(f"Array : {array}\n{type(array)}")

print(f"Length of set : {len(s)}")
print(f"Length of Array : {array.size}")
```

```python
# Create a numpy array with elements from 1 to 100

li=[i for i in range(1,101,1)]
arr=np.array(li)
print(arr)
```

```python
# print all the elements from 1 to 100 which are divisible by 3

li = [item for item in range(1, 101, 1) if item % 3 == 0]
array = np.array(li)

print(f"Array : {array}")
```

```python
# print array with 3 rows and 5 columns with all the elements as zero

rows, columns, data = 3, 5, 0

li = [[data for i in range(columns)] for i in range(rows)]
array = np.array(li)

print(f"Array\n{array}")
```
