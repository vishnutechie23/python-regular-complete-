1 Arithmetic Operations : 
Arithmetic operations perform mathematical calculations on NumPy arrays.

Notice : NumPy performs the operation on every element.
Arithmetic operators

    Operator   Meaning      	Example
    +	Addition	        arr + 5
    -	Subtraction	        arr - 5
    *	Multiplication	    arr * 2
    /	Division	        arr / 2
    //	Floor division	    arr // 3
    %	Modulus	            arr % 3
    **	Power	            arr ** 2

```python
import numpy as np
arr = np.array([10, 20, 30, 40, 50])
print(arr + 5)
print(arr - 5)
print(arr * 2)
print(arr / 2)

# Try all operators:
arr = np.array([10, 20, 30, 40, 50])
print(arr // 3)
print(arr % 3)
print(arr ** 2)

# Important: / normally produces floating-point results, even when both inputs are integers.
```

2 Operations Between Two Arrays : 
You can perform arithmetic operations between two arrays of the same shape.

```python

a = np.array([10, 20, 30, 40])
b = np.array([2, 4, 5, 8])
print(a + b)
print(a - b)
print(a * b)
print(a / b)

# Each element of a is operated on with the corresponding element of b.

# This is called element-wise operation.
# Important: * vs @
# For arrays, * means element-wise multiplication.
a = np.array([[1, 2],
              [3, 4]])
b = np.array([[5, 6],
              [7, 8]])
print(a * b)
print(a @ b)

# - * → element-wise multiplication.
# - @ → matrix multiplication.
# We'll explore matrix multiplication in more detail in the Linear Algebra module.
```
