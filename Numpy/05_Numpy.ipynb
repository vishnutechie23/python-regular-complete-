1 Arithmetic Operations : 
Arithmetic operations perform mathematical calculations on NumPy arrays.

Notice : NumPy performs the operation on every element.
Arithmetic operators

    Operator   Meaning      	Example
    +	Addition	        arr + 5
    -	Subtraction	        arr - 5
    *	Multiplication	    arr * 2
    /	Division	        arr / 2
    //	Floor division	    arr // 3
    %	Modulus	            arr % 3
    **	Power	            arr ** 2

```python
import numpy as np
arr = np.array([10, 20, 30, 40, 50])
print(arr + 5)
print(arr - 5)
print(arr * 2)
print(arr / 2)

# Try all operators:
arr = np.array([10, 20, 30, 40, 50])
print(arr // 3)
print(arr % 3)
print(arr ** 2)

# Important: / normally produces floating-point results, even when both inputs are integers.
```

2 Operations Between Two Arrays : 
You can perform arithmetic operations between two arrays of the same shape.

```python

a = np.array([10, 20, 30, 40])
b = np.array([2, 4, 5, 8])
print(a + b)
print(a - b)
print(a * b)
print(a / b)

# Each element of a is operated on with the corresponding element of b.

# This is called element-wise operation.
# Important: * vs @

# For arrays, * means element-wise multiplication.
a = np.array([[1, 2],
              [3, 4]])
b = np.array([[5, 6],
              [7, 8]])
print(a * b)
print(a @ b)

# - * → element-wise multiplication.
# - @ → matrix multiplication.
# We'll explore matrix multiplication in more detail in the Linear Algebra module.
```

### Broadcasting
Broadcasting allows NumPy to perform operations between arrays of different shapes, provided their dimensions are compatible.

```python
# For example:
arr = np.array([10, 20, 30, 40])
print(arr + 5)
# Output : [15 25 35 45]
# Here, NumPy behaves as if the scalar 5 is applied to every element.

# Broadcasting with 1D and 2D arrays
arr = np.array([[10, 20, 30],
                [40, 50, 60]])
b = np.array([1, 2, 3])
print(arr + b)

# Output:
# [[11 22 33]
#  [41 52 63]]


# Why does this work?
# The shapes are:
# arr → (2, 3)
# b   →    (3)

# NumPy aligns dimensions from the right:
# arr → (2, 3)
# b   → (1, 3)

# The dimensions are compatible, so the operation works.
# Broadcasting rules
# Compare dimensions from right to left. Two dimensions are compatible if:
# 1. They are equal, OR
# 2. One of them is 1.
# If neither condition is satisfied, broadcasting fails.
```

```python
# Example 1 – Compatible (np.ones : Fill with 1)
a = np.ones((3, 4))
b = np.ones((4,))
print(a + b)

# Shapes: (3, 4) and (4,) → compatible.
```

```python
# Example 2 – Not compatible
a = np.ones((3, 4))
b = np.ones((3,))
print(a + b)

# Shapes: (3, 4) and (3,) → incompatible because the last dimensions are 4 and 3. 
# This produces a ValueError : ValueError: operands could not be broadcast together with shapes (3,4) (3,)
```

```python
# Broadcasting using np.newaxis
# Sometimes you need to change the shape of an array to make broadcasting work.
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])
result = a[:, np.newaxis] + b
print(result)

# Here:
# - a[:, np.newaxis] changes the shape from (3,) to (3, 1).
# - b has shape (3,).
# - Broadcasting produces a (3, 3) result.

# Syntax	                        Shape
# arr[:, np.newaxis]	            (3, 1) – Column
# arr[np.newaxis, :]	            (1, 3) – Row
```

#### Comparison Operations

    Comparison operations compare array elements and return Boolean values (True or False).

        Comparison operators
        Operator	Meaning
        ==	        Equal to
        !=	        Not equal to
        >	        Greater than
        <	        Less than
        >=	        Greater than or equal to
        <=	        Less than or equal to

```python
arr = np.array([10, 20, 30, 40, 50])
print(arr > 25)
print(arr < 30)
print(arr == 20)
print(arr != 40)
print(arr >= 30)
print(arr <= 20)
```

#### Logical Operations
Logical operations combine multiple conditions.

    # Logical operators
    # Operator                      Meaning	                        NumPy symbol
    # AND	                Both conditions true	                    &
    # OR	                Either condition true	                    \|
    # NOT                   Reverse condition	                        ~
    # Logical AND function	Element-wise AND	                    np.logical_and()
    # Logical OR function	Element-wise OR	                        np.logical_or()
    # Logical NOT function	Element-wise NOT	                    np.logical_not()

```python
# Suppose you want employees who:
# - Are younger than 30.
# - Earn more than ₹50,000. You can use & (AND).
condition = ((data['Age'] < 30) &    (data['Monthly_Salary'] > 50000))
result = data[condition]
print(result)

# Example using OR:
condition = ((data['Experience_Years'] > 10) |    (data['Projects_Completed'] > 10))
result = data[condition]
print(result)

# Example using NOT:
condition = ~(data['Age'] < 50)
result = data[condition]
print(result)

# Important: Always put each comparison inside parentheses when using & or |.
# Correct:
# (data['Age'] > 25) & (data['Monthly_Salary'] > 40000)
# Incorrect:
# data['Age'] > 25 & data['Monthly_Salary'] > 40000
```

#### Aggregate Operations
Aggregate operations calculate a single result from multiple array elements.

    Important       aggregate functions
    Function	        Purpose
    np.sum()	        Total
    np.min()	        Minimum
    np.max()	        Maximum
    np.mean()	        Average
    np.median()	        Median
    np.std()	        Standard deviation
    np.var()	        Variance
    np.prod()	        Product of elements
    np.count_nonzero()	Count non-zero elements

```python
arr = np.array([10, 20, 30, 40, 50])
print(np.sum(arr))
print(np.min(arr))
print(np.max(arr))
print(np.mean(arr))
print(np.std(arr))
print(np.var(arr))
```

#### Universal Functions (ufuncs)
A universal function (ufunc) is a NumPy function that performs an operation element by element, efficiently.

    Common ufuncs
    Function	            Purpose	                Example
    np.sqrt()	            Square root	            np.sqrt(arr)
    np.square()	            Square	                np.square(arr)
    np.abs()	            Absolute value	        np.abs(arr)
    np.exp()	            Exponential	            np.exp(arr)
    np.log()	            Natural logarithm	    np.log(arr)
    np.log10()	            Base-10 logarithm	    np.log10(arr)
    np.sin()	            Sine	                np.sin(arr)
    np.cos()	            Cosine	                np.cos(arr)
    np.floor()	            Round down	            np.floor(arr)
    np.ceil()	            Round up	            np.ceil(arr)
    np.round()	            Round values	        np.round(arr, 2)


```python
# Ex 1 :
arr = np.array([1, 4, 9, 16, 25])
print(np.sqrt(arr))

# Ex 2 :
arr = np.array([1.234, 2.567, 3.891])
print(np.round(arr, 2))
print(np.floor(arr))
print(np.ceil(arr))
```

#### Vectorization
Vectorization means applying operations to an entire array instead of manually processing elements using Python loops.

    Why is vectorization useful?
- Less code.
- Usually faster for large numerical arrays.
- Easier to perform mathematical calculations.
- Widely used in Data Analytics and Machine Learning.

```python
# Without vectorization
arr = [10, 20, 30, 40, 50]
result = []
for i in arr:
    result.append(i * 2)
print(result)

# With NumPy vectorization
arr = np.array([10, 20, 30, 40, 50])
result = arr * 2
print(result)

```

#### Conditional Operations – np.where()
    np.where() selects values based on a condition.

    Syntax:
        np.where(condition, value_if_true, value_if_false)

```python
# Example:
marks = np.array([85, 45, 70, 30, 90])
result = np.where(marks >= 50, "Pass", "Fail")
print(result)


# Practical example – Employee salary category
salary = data['Monthly_Salary']
category = np.where(salary >= 50000, "High Salary", "Low Salary")
print(category)


# You can also use np.where() to get the positions where a condition is true:
arr = np.array([10, 25, 30, 45, 50])
result = np.where(arr > 30)
print(result)

# The matching positions are 3 and 4.
```
